```python
import json
```

```python
path  =  '/Users/finny/Documents/Exoplanets/CTOI Creation/Results/Data/TIC 300291839 Exoplanet b.json'

with open(path, 'rb') as file:
    data  =  json.loads(file.read())

data
```

```python
from vespa.transitparams import TransitSignal

host_star = data['host_star']
# Create TransitSignal object for vespa
signal = TransitSignal(
    name = data['name'],
    period = data['period'],
    epoch = data['epoch'],
    duration = data['duration'],
    depth = data['depth'],
    ra = host_star['ra'],
    dec = host_star['dec'],
    Teff = host_star['Teff'],
    logg = host_star['logg'],
    feh = host_star['feh'],
    star_radius = host_star['radius'],
    star_mass = host_star['mass'],
    phot = host_star['photometry']
)

output_path = '/mnt/data/TIC300291839_b.xml'
signal.to_xml(output_path)
```
