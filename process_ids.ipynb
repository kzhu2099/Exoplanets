```python
import pandas
```

```python
parent_folder = 'CTOI Creation'
```

```python
toi_data = pandas.read_csv('https://exofop.ipac.caltech.edu/tess/download_toi?output=csv')
ctoi_data = pandas.read_csv('https://exofop.ipac.caltech.edu/tess/download_ctoi?output=csv')
koi_data = pandas.read_csv('https://exofop.ipac.caltech.edu/tess/download_koi?sort=koi&output=csv')

toi_data, ctoi_data, koi_data
```

```python
toi_data['type'] = 'TOI'
ctoi_data['type'] = 'CTOI'
koi_data['type'] = 'KOI'

columns = ['TIC ID', 'type']

all_planets = pandas.concat([toi_data[columns], ctoi_data[columns], koi_data[columns]], ignore_index = True, sort = False)
all_planets
```

```python
all_planets = all_planets.drop_duplicates(subset = 'TIC ID', keep = 'first').reset_index(drop = True)
all_planets = all_planets.sort_values(by = 'TIC ID').reset_index(drop = True)

all_planets.to_csv(f'{parent_folder}/Current IDs/all_planets.csv', index = False)

all_planets
```

```python
completed_ids = pandas.read_csv(f'{parent_folder}/Current IDs/completed_tic_ids.csv')['ticid'].astype(int)
known_ids = all_planets['TIC ID'].astype(int)

completed_ids, known_ids
```

```python
data_path = f'{parent_folder}/TIC ID Data/tess_tce_80s.csv'
depth = 0.075 * 1e6
```

```python
data = pandas.read_csv(data_path)
data
```

```python
data = data[~data['ticid'].isin(completed_ids)]
data = data[~data['ticid'].isin(known_ids)]
data = data[data['tce_depth'] < depth] # any depth
data = data.reset_index(drop = True)

data
```

```python
short_data = data[(data['tce_period'] > 10 ** 0.3) & (data['tce_period'] < 10 ** 1)]
short_ids = short_data['ticid'].drop_duplicates().reset_index(drop = True)

short_ids.to_csv(f'{parent_folder}/Current IDs/tic_ids_short_period.csv', index = False)
short_ids
```

```python
long_data = data[(data['tce_period'] > 10 ** 1) & (data['tce_period'] < 10 ** 1.7)]
long_ids = long_data['ticid'].drop_duplicates().reset_index(drop = True)

long_ids.to_csv(f'{parent_folder}/Current IDs/tic_ids_long_period.csv', index = False)
long_ids
```
